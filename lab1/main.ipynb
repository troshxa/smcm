```python
import numpy as np; import matplotlib.pyplot as plt;
#Exponential growth dN/dt=rN

n0 = 5#starting population
r = 0.6#growth const
t0=0#starting time
t_max=10#max simulation time
dt=0.2#time step

t = np.arange(t0, t_max + dt, dt)
steps = len(t)

n =np.zeros(steps)
n[0] = n0

for i in range(steps-1):
    dndt = n[i]*r
    n[i+1] = n[i] +dndt*dt #euler's method

plt.plot(t, n, marker = 'o', markersize = 2, label = "Exponential Growth")
plt.xlabel("time(t)")
plt.ylabel("number(n)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np; import matplotlib.pyplot as plt;

#Newton's law of cooling dT/dt = -k(T-A)

temp0 =170#starting temp
a = 65#equilibrium temp
k = 0.1#cooling const
t_max = 30#simulation time limit
t0= 0#starting time
dt=0.2#time step

t = np.arange(t0, t_max + dt, dt)
steps = len(t)

temp =np.zeros(steps)
temp[0] = temp0

for i in range(steps-1):
    dtempdt = -(temp[i]-a)*k
    temp[i+1] = temp[i] +dtempdt*dt

plt.plot(t, temp, marker = 'o', markersize = 1, label = "Newton's law of cooling")
plt.xlabel("time(t)")
plt.ylabel("temperature(T)")
plt.grid(True)
plt.legend()
plt.show()

```

```python
import numpy as np; import matplotlib.pyplot as plt;
#Lorenz Attractor
#dx/dt = sigma(y-x)
#dy/dt =x(p-z)-y
#dz/dt = xy -bz

#standart system consts 
sigma =10
p = 28
b = 8/3
t_max = 50#simulation time limit
t0 = 0#starting time
x0=0.3# starting x
y0=0.4 #starting y
z0=0.1#starting z
dt = 0.007#time step

t = np.arange(t0, t_max + dt, dt)
steps = len(t)

x =np.zeros(steps)
x[0] = x0

y =np.zeros(steps)
y[0] = y0

z =np.zeros(steps)
z[0] = z0

for i in range(steps-1):
    dxdt = sigma*(y[i]-x[i])
    dydt = x[i]*(p-z[i]) - y[i]
    dzdt = x[i]*y[i] - b*z[i]

    x[i+1] = x[i] + dxdt*dt
    y[i+1] = y[i] + dydt*dt
    z[i+1] = z[i] + dzdt*dt
plt.plot(x, z, lw=0.5, color='purple')
plt.xlabel('X')
plt.ylabel('Z')
plt.title('Lorenz Attractor: X-Z Projection')
plt.grid(True)
plt.show()
```
